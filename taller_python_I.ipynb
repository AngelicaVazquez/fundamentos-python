### 01 · Conversor de temperatura
Pide una temperatura en grados Celsius con `input()` y muestra su equivalente en Fahrenheit **con dos decimales**.

Fórmula: `F = C × 9/5 + 32`

**Salida esperada** (si escribes `21.5`):
```
21.5 °C son 70.70 °F
```
*Practicas: input, float, f-strings.*

````python
celsius = float(input("Introduce la temperatura en grados Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius:.2f} ºC son {fahrenheit:.2f} ºF")
````
